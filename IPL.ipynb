```python
import pandas as pd 
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
#loading the ipl matches dataset
ipl=pd.read_csv("ipl.csv")
```

```python
#first five recodes of the dataset
ipl.head()
```

```python
#lokking at number of rows and column in the dataset
ipl.shape
```

```python
#Getting the frequence man of the match
ipl["player_of_match"].value_counts()
```

```python
#top 10 player man of the match
ipl["player_of_match"].value_counts()[0:10]
```

```python
list(ipl["player_of_match"].value_counts()[0:10].keys())
```

```python
#Making a bar-plot for the top 7 player
plt.figure(figsize=(8,5))
plt.bar(list(ipl["player_of_match"].value_counts()[0:7].keys()),list(ipl["player_of_match"].value_counts()[0:7]),color="blue")
plt.show()
```

```python
#the frequency of result column
ipl["result"].value_counts()
```

```python
#finding out of the numbers of toss win w.r.t each team
ipl["toss_winner"].value_counts()
```

```python
#Extracting the records where a team won batting first
batting_first=ipl[ipl["win_by_runs"]!=0]
```

```python
#looking at the head
batting_first.head()
```

```python
#Making Histogram 
plt.figure(figsize=(5,5))
plt.hist(batting_first["win_by_runs"])
plt.title("Distubated of Run")
plt.xlabel("Run")
plt.show()
```

```python
#finding out the number of wins w.r.t each team  after batting first
batting_first["winner"].value_counts()
```

```python
#Making a bar plot for top of 3 team with most win after batting first
plt.figure(figsize=(15,10))
plt.bar(list(batting_first["winner"].value_counts()[0:5].keys()),list(batting_first["winner"].value_counts()[0:5]),color=["blue","orange","yellow","brown","red"])
plt.show()
```

```python
#Making apie chart
plt.figure(figsize=(10,10))
plt.pie(list(batting_first["winner"].value_counts()), labels = list(batting_first["winner"].value_counts().keys()),autopct='%0.1f%%')
plt.show()
```

```python
#Extracting those records where a team has won after batting second
batting_second=ipl[ipl["win_by_wickets"]!=0]
```

```python
batting_second.head(10)
```

```python
#Making histogram for frequency of w.r.t number of wickets
plt.figure(figsize=(5,5))
plt.hist(batting_second["win_by_wickets"], bins=20)
plt.title("Distubated by wickets")
plt.show()
```

```python
#finding the frequency of number of win w.r.t each time after batting second
batting_second["winner"].value_counts()
```

```python
#making a bar plot for top 5 team  win after second batting
plt.figure(figsize=(15,10))
plt.bar(list(batting_second["winner"].value_counts()[0:5].keys()),list(batting_second["winner"].value_counts()[0:5]),color=["purple","orange","blue","yellow","red"])
plt.show()
```

```python
#making a pie chart 
plt.figure(figsize=(10,10))
plt.pie(list(batting_second["winner"].value_counts()), labels=list(batting_second["winner"].value_counts().keys()),autopct='%0.1f%%')
plt.show()
```
